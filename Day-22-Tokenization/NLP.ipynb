| Concept      | Meaning                               | Example                                     |
| ------------ | ------------------------------------- | ------------------------------------------- |
| Tokenization | Breaking text into pieces             | `"I love Python"` → `["I","love","Python"]` |
| Vocabulary   | Collection/mapping of known tokens    | `"Python" → 4`                              |
| Token ID     | Numerical identifier of a token       | `"Python" → 4`                              |
| Embedding    | Numerical vector representing a token | `Python → [0.2, -0.5, 0.8, ...]`            |


```python
# Small Practical Exercise

import torch
import torch.nn as nn

# 1. Vocabulary
vocabulary = {
    "I" : 0,
    "love" : 1,
    "Python" : 2,
    "AI" : 3
}

# 2. Sentence
sentence = ['I', 'love', 'Python', 'AI']

# 3. Convert token to IDs
token_ids = [vocabulary[token] for token in sentence]
print("Token IDs: ", token_ids)

# 4. Convert to tensor
token_ids = torch.tensor(token_ids)

# 5. Create embedding layer
embedding = nn.Embedding(
    num_embeddings=4,
    embedding_dim=5
)

# 6. Get embedding
vectors = embedding(token_ids)

print("\nEmbedding: ")
print(vectors)

print("\nShape: ", vectors.shape)
```
